```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")


plt.figure(figsize=(8, 6))

sns.violinplot(
    data=tips,
    x="day", 
    y="total_bill",
    palette="magma",  
    inner="box"       
)

plt.xlabel("day")
plt.ylabel("total_bill")
plt.title("") 

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
tips.day = tips.day.replace({"Fri":"Mon","Sat":"Tue","Sun":"Wed","Thur":"Thurs"})

sns.violinplot(data=tips, x="day", y="total_bill", 
               order=["Mon","Tue","Wed","Thurs"], 
               palette="coolwarm", inner="box")

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ['Pants', 'Shirt', 'Accessories', 'Jeans', 'Top']
sales = [500, 630, 280, 840, 1000]

plt.figure(figsize=(6,4))
plt.barh(categories, sales, color='orange')

plt.xlabel('sales')
plt.ylabel('Product')
plt.title('Graph')

plt.show()
```

```python
import matplotlib.pyplot as plt


days = ['Mon', 'Tues', 'Wed', 'Thurs', 'Fri', 'Sat']
sales = [32, 42, 17, 44, 14, 29]  

plt.figure(figsize=(5,3))
plt.plot(days, sales, color='blue', marker='o', linewidth=2)

plt.xlabel('days')
plt.ylabel('count')
plt.title('Graph')

plt.ylim(10, 60) 
plt.show()
```

```python
import matplotlib.pyplot as plt
subjects = ["python", "sql", "power BI"]
students = [20,34,56]
plt.figure(figsize=(7,6))
plt.pie(
    students,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90,
    explode= (0.1,0,0)
)
plt.title("student in different Courses")
plt.savefig("student.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(6,6))

house_x = [1, 4, 4, 1, 1]
house_y = [1, 1, 4, 4, 1]
plt.plot(house_x, house_y, 'b-', linewidth=3
         
roof_x = [1, 2.5, 4, 1]
roof_y = [4, 6, 4, 4]
plt.plot(roof_x, roof_y, 'r-', linewidth=3)

door_x = [2, 3, 3, 2, 2]
door_y = [1, 1, 2.5, 2.5, 1]
plt.plot(door_x, door_y, 'g-', linewidth=2)

window_x = [1.2, 1.8, 1.8, 1.2, 1.2]
window_y = [2.5, 2.5, 3.1, 3.1, 2.5]
plt.plot(window_x, window_y, 'purple', linewidth=2)

theta = np.linspace(0, 2*np.pi, 100)
circle_radius = 0.3
circle_center_x = 2.5  # center of roof
circle_center_y = 5    # center of roof
circle_x = circle_center_x + circle_radius * np.cos(theta)
circle_y = circle_center_y + circle_radius * np.sin(theta)
plt.plot(circle_x, circle_y, 'orange', linewidth=2)


plt.title("My House", fontsize=14)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid(True, linestyle='--', alpha=0.5)
plt.axis('equal')
plt.xlim(0, 6)
plt.ylim(0, 7)

plt.show()
```
